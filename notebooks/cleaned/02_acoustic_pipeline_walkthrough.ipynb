# MindText - acoustic-feature pipeline walkthrough (SYNTHETIC data)

Synthetic Gaussian features with a built-in signal; **not E-DAIC**. Shows the protocol: tune on train, threshold & model choice on dev, test once, with the all-positive baseline printed next to the scores. Outputs not stored.

```python
import tempfile, pathlib
from mindtext.config import Config
from mindtext.synthetic import make_tabular_dataset
from mindtext.training import run_experiment
from mindtext.reporting import format_report_text

work = pathlib.Path(tempfile.mkdtemp())
make_tabular_dataset(seed=0).to_csv(work/'tab.csv', index=False)
```

```python
cfg = Config.from_yaml('../../configs/acoustic.yaml') if pathlib.Path('../../configs/acoustic.yaml').exists() else Config.from_yaml('configs/acoustic.yaml')
cfg.output.reports_dir = str(work/'reports'); cfg.output.artifacts_dir = str(work/'artifacts')
res = run_experiment(cfg, data_path=work/'tab.csv', log=lambda m: None)
print(format_report_text(res.report))
```

```python
sel = next(m for m in res.report['models'] if m['name'] == res.report['selected_model'])
print('threshold chosen on dev:', sel['threshold'])
print('test 95% CI:', sel['test_ci'])
```
